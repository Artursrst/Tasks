```python
#Уровень 6
# 1 Сделайте функцию, которая параметром будет принимать букву и проверять, 
# это буква кириллицы или латиницы.

def check(letter):
    cyrillic_str = "абвгдеёжзийклмнопрстуфхцчшщъыьэюяАБВГДЕЁЖЗИЙКЛМНОПРСТУФХЦЧШЩЪЫЬЭЮЯ"
    latin_str = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ"
    if letter in cyrillic_str:
        return 'cyrillic'
    if letter in latin_str:
        return 'latin'
    else:
        return 'Буква не является латиницей или кирилицей'
```

```python
#2 Сделайте функцию, которая параметром будет принимать список и 
# перемешивать элементы этого списка в случайном порядке.
import random

list_1 = [1, 2, 3, 9, 4, 5]

def shake_list(lst):
    indexes = []
    result = lst.copy()
    i = random.randint(0, len(lst) - 1)
    for e in lst:        
        while i in indexes:
            i = random.randint(0, len(lst) - 1)
        indexes.append(i)

        result[i] = e
    
    return result

print(shake_list(list_1))   
```

```python
#3 Дан текст со словами. Запишите все слова этого текста в специальный словарь. 
# Ключами в этом словаре будут буквы, а значениями - списки слов, 
# начинающихся на эти буквы.

text = 'Ключами в этом словаре будут буквы а значениями списки слов'
def text_dict(text):
    words = text.split(' ')
    result = {}
    for word in words:
        if result.get(word[0]):
            result[word[0]] = result[word[0]] + ' ' + word
        else:
            result[word[0]] = word

    return result

print(text_dict(text))
```

```python
#4 Сделайте функцию, которая параметром будет принимать число, 
# а возвращать список его делителей, являющихся простыми числами.

def func(n):
    divisors = []
    d = 2

    while d * d <= n:
        if n % d == 0:
            divisors.append(d)
            while n % d == 0:
                n //= d
        d += 1

    if n > 1:
        divisors.append(n)

    return divisors
```

```python
#5 Сделайте функцию, которая параметром будет принимать слово и 
# возвращать список его слогов.

def split_word(word):
    vowels = "аеёиоуыэюяАЕЁИОУЫЭЮЯ"

    vowel_indices = [i for i, char in enumerate(word) if char in vowels]
    
    if len(vowel_indices) <= 1:
        return [word] if word else []
    
    syllables = []
    start = 0
    
    for i in range(len(vowel_indices) - 1):
        curr_v = vowel_indices[i]
        next_v = vowel_indices[i + 1]
        
        consonants_between = next_v - curr_v - 1
        
        if consonants_between == 0:
            split_pos = curr_v + 1
        elif consonants_between == 1:
            split_pos = curr_v + 1
        else:
            if word[curr_v + 1].lower() in "йьъ":
                split_pos = curr_v + 2
            else:
                split_pos = curr_v + 2
                
        syllables.append(word[start:split_pos])
        start = split_pos
        
    syllables.append(word[start:])
    
    return syllables

print(split_word('молочко'))
```

```python
#Уровень 10
# 3
'''Дан произвольный двухмерный список:

[
	[11, 12, 13, 14, 15],
	[21, 22, 23, 24, 25],
	[31, 32, 33, 34, 35],
	[41, 42, 43, 44, 45],
	[51, 52, 53, 54, 55],
]
Выполните его транспонирование:

[
	[11, 21, 31, 41, 51],
	[12, 22, 32, 42, 52],
	[13, 23, 33, 43, 53],
	[14, 24, 34, 44, 54],
	[15, 25, 35, 45, 55],
]'''

import copy

m = [
	[11, 12, 13, 14, 15],
	[21, 22, 23, 24, 25],
	[31, 32, 33, 34, 35],
	[41, 42, 43, 44, 45],
	[51, 52, 53, 54, 55],
]

t=copy.deepcopy(m)

for i in range(len(m)):
    for j in range(len(m[0])):
        t[j][i] = m[i][j]

t
```
